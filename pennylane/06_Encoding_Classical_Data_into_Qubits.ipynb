## Real-World Applications & Modern Use Cases

Feature mapping and quantum embedding are essential for:
- **Medical Image Diagnostics:** Encoding high-dimensional histological or radiological feature vectors into qubit state spaces for classification.
- **High-Energy Physics Event Filtering:** Mapping sensor signals from particle colliders (such as the Large Hadron Collider) into quantum Hilbert spaces for anomaly detection.


```python
import pennylane as qml
import numpy as np

dev = qml.device("default.qubit", wires=3)

@qml.qnode(dev)
def angle_embedding_circuit(features):
    qml.AngleEmbedding(features, wires=range(3), rotation='Y')
    return qml.probs(wires=range(3))

sample_features = [0.5, 1.2, 2.1]
probabilities = angle_embedding_circuit(sample_features)

print("Input features:     ", sample_features)
print("State probabilities:", np.round(probabilities, 3))
print("\nCircuit structure:")
print(qml.draw(angle_embedding_circuit)(sample_features))
```
