# Notebook 11: Max-Cut Optimization with PennyLane QAOA

This notebook uses the `qml.qaoa` module to formulate and solve graph Max-Cut problems.

---

## Learning Objectives
1. Define problem graphs with NetworkX.
2. Generate cost and mixer Hamiltonians using `qaoa.maxcut`.
3. Construct QAOA layers.


---
## Real-World Applications & Modern Use Cases

Combinatorial optimization with QAOA is applied in industry to:
- **Telecommunication Network Design:** Optimizing fiber-optic link placement and mobile tower cell coverage under complex topological graph constraints.
- **Power Grid Load Balancing:** Computing optimal power distribution paths across electrical smart grids during peak load demands.


```python
import networkx as nx
import pennylane as qml
from pennylane import qaoa

G = nx.Graph([(0, 1), (1, 2), (2, 0)])
cost_h, mixer_h = qaoa.maxcut(G)

dev = qml.device("default.qubit", wires=3)

def qaoa_layer(gamma, beta):
    qaoa.cost_layer(gamma, cost_h)
    qaoa.mixer_layer(beta, mixer_h)

@qml.qnode(dev)
def qaoa_circuit(params):
    for w in range(3):
        qml.Hadamard(wires=w)
    qaoa_layer(params[0], params[1])
    return qml.expval(cost_h)

output = qaoa_circuit([0.5, 0.3])
print(f"QAOA Cost expectation: {output:.4f}")
```
