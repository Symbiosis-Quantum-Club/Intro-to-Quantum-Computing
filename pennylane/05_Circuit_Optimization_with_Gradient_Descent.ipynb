# Notebook 05: Circuit Optimization with Gradient Descent

This notebook demonstrates parameter optimization using gradient descent.

---

## Learning Objectives
1. Define variational loss functions.
2. Optimize circuit parameters using `qml.AdamOptimizer`.
3. Track and evaluate loss convergence.


---
## Real-World Applications & Modern Use Cases

Gradient-based circuit optimization drives:
- **Algorithmic Trading & Arbitrage:** Tuning quantum variational circuits to identify non-linear arbitrage opportunities across financial order books.
- **Quantum State Preparation:** Compiling target multi-qubit states with minimal circuit depth for fault-tolerant algorithmic initialization.


```python
import pennylane as qml
from pennylane import numpy as pnp

dev = qml.device("default.qubit", wires=1)

@qml.qnode(dev)
def parametric_circuit(params):
    qml.RX(params[0], wires=0)
    qml.RY(params[1], wires=0)
    return qml.expval(qml.PauliZ(0))

def cost_fn(params):
    return parametric_circuit(params)

params = pnp.array([0.1, 0.2], requires_grad=True)
opt = qml.AdamOptimizer(stepsize=0.1)

print("Optimization environment ready.")
```

---
## Section 1: Optimization Loop Execution


```python
history = []
for step in range(25):
    params, current_cost = opt.step_and_cost(cost_fn, params)
    history.append(current_cost)

print(f"Initial Cost:         {history[0]:.4f}")
print(f"Final Optimized Cost: {history[-1]:.4f} (Target: -1.0000)")
print("Optimization converged successfully.")
```
