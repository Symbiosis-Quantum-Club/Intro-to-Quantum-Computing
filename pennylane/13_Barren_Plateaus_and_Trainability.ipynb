# Notebook 13: Barren Plateaus and Trainability in QML

This notebook explores the phenomenon of vanishing gradients (barren plateaus) in parameterized quantum circuits.

---

## Learning Objectives
1. Analyze gradient variance scaling across qubit registers.
2. Understand circuit depth and trainability constraints.


---
## Real-World Applications & Modern Use Cases

Understanding barren plateau phenomena is vital for:
- **Scalable Quantum Architecture Engineering:** Diagnosing and mitigating vanishing gradients in deep variational circuits via local cost functions and layer-by-layer pre-training.
- **Industry Feasibility Benchmarking:** Establishing theoretical and empirical trainability boundaries before scaling quantum neural networks to hundreds of qubits.


```python
import pennylane as qml
from pennylane import numpy as pnp

def measure_gradient_variance(num_qubits=2, samples=10):
    dev = qml.device("default.qubit", wires=num_qubits)
    
    @qml.qnode(dev)
    def circuit(weights):
        qml.RandomLayers(weights, wires=range(num_qubits))
        return qml.expval(qml.PauliZ(0))
        
    grad_fn = qml.grad(circuit)
    grads = []
    for _ in range(samples):
        weights = pnp.random.random((2, num_qubits), requires_grad=True)
        grads.append(grad_fn(weights)[0, 0])
    return float(pnp.var(grads))

v2 = measure_gradient_variance(2)
v3 = measure_gradient_variance(3)

print(f"Gradient Variance (2 qubits): {v2:.6f}")
print(f"Gradient Variance (3 qubits): {v3:.6f}")
```
