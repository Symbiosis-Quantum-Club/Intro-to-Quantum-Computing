## Real-World Applications & Modern Use Cases

Quantum chemistry simulation is one of the most promising near-term quantum applications:
- **Clean Energy Catalyst Design:** Simulating nitrogenase and transition-metal complexes to develop efficient synthetic catalysts for fertilizer production and carbon capture.
- **Next-Generation Battery Chemistry:** Modeling lithium-ion electrolyte degradation and cathode phase changes beyond the limits of classical density functional theory.


```python
import pennylane as qml
from pennylane import numpy as np

symbols = ["H", "H"]
coordinates = np.array([0.0, 0.0, -0.6614, 0.0, 0.0, 0.6614])

H, qubits = qml.qchem.molecular_hamiltonian(symbols, coordinates)
print(f"Required Qubits for H2 Molecule: {qubits}")
print("Molecular Hamiltonian representation:")
print(H)
```
