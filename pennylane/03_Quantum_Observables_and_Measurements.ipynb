## Real-World Applications & Modern Use Cases

Statistical quantum observables are central to:

- **Classical Shadow Tomography:** Extracting large numbers of quantum observable predictions from minimal physical measurement shots in quantum materials research.


```python
import pennylane as qml
import numpy as np

# Device configuration
dev = qml.device("default.qubit", wires=2)
print("Measurement evaluation device loaded.")
```

---
## Section 1: Returning Multiple Quantum Statistics


```python
@qml.qnode(dev)
def measurement_circuit(theta):
    qml.RY(theta, wires=0)
    qml.CNOT(wires=[0, 1])
    return (
        qml.expval(qml.PauliZ(0)),
        qml.var(qml.PauliZ(0)),
        qml.probs(wires=[0, 1])
    )

exp_val, variance, probs = measurement_circuit(np.pi / 3)

print(f"Expectation value <Z0>: {exp_val:.4f}")
print(f"Variance Var(Z0):       {variance:.4f}")
print("Basis probabilities (|00>, |01>, |10>, |11>):", np.round(probs, 3))
```
