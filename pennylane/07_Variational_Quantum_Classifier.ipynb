# Notebook 07: Variational Quantum Classifier

This notebook constructs a variational quantum classifier (VQC) using strongly entangling layers.

---

## Learning Objectives
1. Formulate the VQC architecture.
2. Apply `qml.StronglyEntanglingLayers`.
3. Compute predictions for sample feature vectors.


---
## Real-World Applications & Modern Use Cases

Variational quantum classifiers (VQCs) are actively explored in:
- **Fraud Detection & Financial Security:** Identifying subtle, non-linear transaction patterns indicative of financial fraud using strongly entangling layers.
- **Genomic Sequence Classification:** Classifying DNA and RNA sequence motifs associated with rare genetic disorders using variational quantum classifiers.


```python
import pennylane as qml
from pennylane import numpy as pnp

num_qubits = 2
dev = qml.device("default.qubit", wires=num_qubits)

@qml.qnode(dev)
def vqc(weights, features):
    qml.AngleEmbedding(features, wires=range(num_qubits))
    qml.StronglyEntanglingLayers(weights, wires=range(num_qubits))
    return qml.expval(qml.PauliZ(0))

shape = qml.StronglyEntanglingLayers.shape(n_layers=2, n_wires=num_qubits)
weights = pnp.random.random(shape, requires_grad=True)
sample_x = pnp.array([0.5, -0.3])

pred = vqc(weights, sample_x)
print(f"Sample prediction: {pred:.4f}")
```
