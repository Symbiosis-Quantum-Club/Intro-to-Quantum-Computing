# Notebook 04: Quantum Gradients and the Parameter-Shift Rule

This notebook demonstrates the parameter-shift rule for evaluating exact analytic derivatives of quantum circuits.

---

## Learning Objectives
1. Understand the analytic parameter-shift formula:
   $$\frac{df}{d\theta} = \frac{f(\theta + s) - f(\theta - s)}{2\sin(s)}$$
2. Compute circuit derivatives using `qml.grad`.
3. Verify quantum gradients against analytic calculus.


---
## Real-World Applications & Modern Use Cases

Analytic gradient evaluation powers:
- **Hardware-Compatible Analytic Gradients:** Calculating exact mathematical derivatives directly on physical QPUs without the numerical instabilities of finite differences.
- **Quantum Optimal Control:** Pulse-level pulse shaping and quantum gate compilation optimized via gradient backpropagation through quantum operations.


```python
import pennylane as qml
from pennylane import numpy as pnp

dev = qml.device("default.qubit", wires=1)

@qml.qnode(dev, diff_method="parameter-shift")
def circuit(x):
    qml.RX(x, wires=0)
    return qml.expval(qml.PauliZ(0))

print("Parameter-shift QNode constructed.")
```

---
## Section 1: Computing Analytic Quantum Gradients


```python
grad_function = qml.grad(circuit)
angle = pnp.array(0.543, requires_grad=True)

computed_gradient = grad_function(angle)
theoretical_gradient = -pnp.sin(angle)

print(f"Angle:                 {float(angle):.4f} rad")
print(f"Computed Gradient:     {float(computed_gradient):.4f}")
print(f"Theoretical Gradient:  {float(theoretical_gradient):.4f}")
assert pnp.isclose(float(computed_gradient), float(theoretical_gradient)), "Gradients must match theory."
print("Analytic parameter-shift gradient verified successfully.")
```
