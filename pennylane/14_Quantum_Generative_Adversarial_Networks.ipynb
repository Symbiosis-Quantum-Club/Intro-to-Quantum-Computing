This notebook demonstrates the architecture of Quantum GANs (QGAN).

## Learning Objectives

## Real-World Applications & Modern Use Cases

Quantum generative models are deployed in:
- **Synthetic Financial Market Generation:** Generating realistic stock price and volatility distributions for stress-testing bank portfolios without leaking private customer data.
- **Quantum State Reconstruction:** Learning unknown quantum state distributions from experimental measurement data in quantum laboratory characterization.


```python
import pennylane as qml
from pennylane import numpy as pnp

dev = qml.device("default.qubit", wires=2)

@qml.qnode(dev)
def generator(weights):
    qml.RY(weights[0], wires=0)
    qml.RY(weights[1], wires=1)
    qml.CNOT(wires=[0, 1])
    return qml.probs(wires=range(2))

init_weights = pnp.array([0.5, 0.8], requires_grad=True)
distribution = generator(init_weights)

print("Generated Probability Distribution:", pnp.round(distribution, 3))
print("\nGenerator Layout:")
print(qml.draw(generator)(init_weights))
```
