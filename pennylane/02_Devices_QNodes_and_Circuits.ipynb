# Notebook 02: Constructing Quantum Nodes (QNodes)

In PennyLane, quantum circuits are defined as Python functions decorated with `@qml.qnode(dev)`.

---

## Learning Objectives
1. Construct and execute a basic QNode.
2. Evaluate parameterized rotation gates.
3. Draw circuits using `qml.draw`.


---
## Real-World Applications & Modern Use Cases

Quantum Nodes (QNodes) serve as the computational interface for:
- **Differentiable Quantum Programming:** Integrating quantum nodes directly into PyTorch/Autograd computational graphs as custom gradient-aware layers.
- **Quantum Sensing Parameter Optimization:** Optimizing RF pulse phase and duration sequences using gradient descent on observable expectation values.


```python
import pennylane as qml
import numpy as np

# Allocate 2 wires on default.qubit
dev = qml.device("default.qubit", wires=2)
print("Device initialized.")
```

---
## Section 1: Defining the QNode

A QNode binds a quantum circuit function to a specific execution device using the `@qml.qnode` decorator.


```python
@qml.qnode(dev)
def parametric_circuit(param):
    # Apply single-qubit parameterized rotation
    qml.RX(param, wires=0)
    # Apply entangling gate
    qml.CNOT(wires=[0, 1])
    # Return Pauli-Z expectation value on wire 1
    return qml.expval(qml.PauliZ(1))

print("QNode created successfully.")
```

---
## Section 2: Executing and Drawing the Circuit


```python
angle = np.pi / 4
expectation_val = parametric_circuit(angle)

print(f"Expectation value at angle {angle:.4f} rad: {expectation_val:.4f}")
print("\nASCII Circuit Diagram:")
print(qml.draw(parametric_circuit)(angle))
```
