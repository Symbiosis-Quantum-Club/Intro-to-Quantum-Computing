## Section 1: Circuit Assembly


```python
from qiskit import QuantumCircuit
from qiskit.primitives import StatevectorSampler

# Step 1: Initial superposition
qc = QuantumCircuit(2)
qc.h([0, 1])
qc.barrier()

# Step 2: Phase oracle targeting state |11>
qc.cz(0, 1)
qc.barrier()

# Step 3: Diffusion operator (H -> X -> CZ -> X -> H)
qc.h([0, 1])
qc.x([0, 1])
qc.cz(0, 1)
qc.x([0, 1])
qc.h([0, 1])
qc.barrier()

# Step 4: Measurement
qc.measure_all()

print("Complete 2-Qubit Grover Search Circuit:")
print(qc.draw(output='text'))
```

---
## Section 2: Sampling and Amplitude Amplification Verification


```python
sampler = StatevectorSampler()
job = sampler.run([qc], shots=1000)
counts = job.result()[0].data.meas.get_counts()

print("Search Measurement Counts (Target: '11'):")
print(counts)
```
