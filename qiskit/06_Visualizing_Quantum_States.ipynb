# Notebook 06: Visualizing Quantum States

This notebook presents standard visualization tools for single- and multi-qubit states using Qiskit and Matplotlib.

---

## Learning Objectives
1. Project single-qubit states onto the 3D Bloch sphere.
2. Render multi-qubit density matrices using State City 3D bars.
3. Compare text, LaTeX, and Matplotlib circuit visualization styles.


---
## Real-World Applications & Modern Use Cases

Quantum state visualization tools are utilized in:
- **Quantum State Tomography:** Verifying experimental physical quantum states synthesized in quantum hardware laboratories.
- **Decoherence Diagnostic Systems:** Tracking pure state decay toward the center of the Bloch sphere as thermal noise dephases the qubit.


---
## Section 1: Creating Parametric Target States

We prepare a single-qubit state with both non-trivial amplitude and phase.


```python
import numpy as np
import matplotlib.pyplot as plt
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

qc = QuantumCircuit(1)
qc.h(0)
qc.rz(np.pi / 4, 0)

sv = Statevector.from_instruction(qc)
print("Statevector Data:", sv.data)
```

---
## Section 2: Bloch Sphere Projection

We render the state on the Bloch sphere using `plot_bloch_multivector`.


```python
from qiskit.visualization import plot_bloch_multivector

fig_bloch = plot_bloch_multivector(sv)
plt.show()
```

---
## Section 3: Multi-Qubit State City Representation

For entangled multi-qubit systems, the density matrix $\rho = |\psi\rangle\langle\psi|$ is plotted in 3D showing real and imaginary components.


```python
from qiskit.visualization import plot_state_city

bell_qc = QuantumCircuit(2)
bell_qc.h(0)
bell_qc.cx(0, 1)

sv_bell = Statevector.from_instruction(bell_qc)

fig_city = plot_state_city(sv_bell, title="Bell State Density Matrix (City Plot)")
plt.show()
```

---
## Section 4: Circuit Rendering Styles

We compare circuit drawing engines: text, latex, and matplotlib.


```python
test_circuit = QuantumCircuit(2)
test_circuit.h(0)
test_circuit.cx(0, 1)
test_circuit.measure_all()

print("ASCII Text Render:")
print(test_circuit.draw(output='text'))

fig_circuit = test_circuit.draw(output='mpl')
plt.show()
```
