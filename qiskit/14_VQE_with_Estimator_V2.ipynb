## Section 1: Defining the Hamiltonian and Ansatz


```python
import numpy as np
from qiskit.circuit.library import TwoLocal
from qiskit.primitives import StatevectorEstimator
from qiskit.quantum_info import SparsePauliOp
from scipy.optimize import minimize

# Molecular Hamiltonian representation for H2
hamiltonian = SparsePauliOp.from_list([
    ("II", -1.05),
    ("ZI", 0.39),
    ("IZ", -0.39),
    ("ZZ", -0.01),
    ("XX", 0.18)
])

# Variational ansatz
ansatz = TwoLocal(num_qubits=2, rotation_blocks='ry', entanglement_blocks='cz', reps=1)
estimator = StatevectorEstimator()

def cost_function(params):
    job = estimator.run([(ansatz, hamiltonian, params)])
    return float(job.result()[0].data.evs)

init_params = np.zeros(ansatz.num_parameters)
res = minimize(cost_function, init_params, method='COBYLA')

print(f"Optimized Ground State Energy: {res.fun:.4f} Hartree")
print(f"Optimization Status: {res.message}")
```
