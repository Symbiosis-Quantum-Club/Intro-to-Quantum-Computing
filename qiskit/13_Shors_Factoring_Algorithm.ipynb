## Section 1: Classical Reduction and Modular Cycles


```python
import math

N = 15
a = 7

print(f"Goal: Factor N = {N} using base a = {a}")
assert math.gcd(a, N) == 1, "Base must be coprime to N."

# Modular exponentiation sequence
sequence = [(a ** x) % N for x in range(8)]
print("Modular sequence for x in 0..7:", sequence)

# Identify period r where a^r = 1 mod N
r = 4
print(f"Detected period r = {r}")

factor_1 = math.gcd(a ** (r // 2) - 1, N)
factor_2 = math.gcd(a ** (r // 2) + 1, N)

print(f"Extracted prime factors: {factor_1} and {factor_2}")
assert factor_1 * factor_2 == N, "Factors must multiply to N."
```
