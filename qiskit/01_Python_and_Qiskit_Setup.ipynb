## Section 1: Python Programming Basics

In Python, external modules are imported to extend functionality. Variables store values, and objects expose methods accessible via dot notation.
Let us initialize basic Python variables that describe our quantum program.


```python
# Define program metadata using standard Python variable types
framework_name = "Qiskit"
target_qubit_count = 1
is_simulation = True

print(f"Target Framework: {framework_name}")
print(f"Target Qubits:    {target_qubit_count}")
print(f"Simulation Mode:  {is_simulation}")
```

---
## Section 2: Importing Qiskit and Verifying Version

We now import the `qiskit` package and check its version. Qiskit 1.0+ introduced a consolidated package architecture that eliminates legacy metapackages.


```python
import qiskit
from qiskit import QuantumCircuit

# Print Qiskit SDK version to confirm 1.x/2.x environment
print("Qiskit SDK Version:", qiskit.__version__)
```

---
## Section 3: Instantiating a QuantumCircuit

A quantum circuit is initialized using the `QuantumCircuit` class. In this baseline example, we allocate 1 quantum bit (qubit) and 1 classical bit for measurement.


```python
# Initialize a circuit with 1 qubit and 1 classical bit
# By default, all qubits are initialized to the computational ground state |0>
qc = QuantumCircuit(1, 1)

print("Quantum circuit object created successfully.")
print(f"Number of qubits:         {qc.num_qubits}")
print(f"Number of classical bits: {qc.num_clbits}")
```

---
## Section 4: Appending Quantum Gates

We now apply physical operations to the circuit. The Pauli-X gate performs a bit-flip operation, mapping $|0\rangle \rightarrow |1\rangle$.


```python
# Apply a Pauli-X gate to qubit 0 (0-indexed)
qc.x(0)

# Add a measurement gate from qubit 0 to classical bit 0
qc.measure(0, 0)

print("Gates successfully appended to the circuit.")
```

---
## Section 5: Inspecting Circuit Properties

We can inspect the computational depth, gate breakdown, and operation counts of the circuit using built-in methods.


```python
# Inspect circuit depth (longest path of non-parallel operations)
circuit_depth = qc.depth()

# Operation counts across the circuit
op_counts = qc.count_ops()

print(f"Circuit Depth: {circuit_depth}")
print(f"Operation Counts: {dict(op_counts)}")
```

---
## Section 6: Visualizing the Circuit Diagram

Qiskit provides both ASCII text-based rendering and publication-quality Matplotlib graphical drawing.


```python
# Render ASCII text representation
print("Text Representation:")
print(qc.draw(output='text'))
```

```python
import matplotlib.pyplot as plt

# Render graphical representation using Matplotlib
fig = qc.draw(output='mpl')
plt.title("Single Qubit Circuit Diagram", fontsize=12)
plt.show()
```

## Section 7: Practice Exercises

### Exercise 1: Multi-Qubit Circuit Construction
Create a 3-qubit circuit with 3 classical bits. Apply an X gate to qubit 0 and qubit 2, leaving qubit 1 in the ground state. Measure all qubits into corresponding classical bits.


```python
# Exercise Solution
ex_qc = QuantumCircuit(3, 3)

# Apply bit flips
ex_qc.x(0)
ex_qc.x(2)

# Measure all qubits
ex_qc.measure([0, 1, 2], [0, 1, 2])

print("Exercise Circuit Diagram:")
print(ex_qc.draw(output='text'))
print(f"Total operations: {dict(ex_qc.count_ops())}")
```
