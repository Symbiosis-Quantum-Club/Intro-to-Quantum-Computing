---
## Section 1: Initializing Registers

The protocol requires 3 qubits and 2 classical bits:
- Qubit 0: Alice's state to transmit ($|\psi\rangle$).
- Qubit 1: Alice's half of the entangled pair.
- Qubit 2: Bob's half of the entangled pair.


```python
from qiskit import QuantumCircuit
import numpy as np

# Allocate 3 qubits and 2 classical bits
qc = QuantumCircuit(3, 2)
print("Registers initialized: 3 qubits, 2 classical bits.")
```

---
## Section 2: Preparing the Secret State on Qubit 0

Alice prepares an arbitrary quantum state using parameterized rotations.


```python
theta = np.pi / 3
qc.ry(theta, 0)
qc.barrier()

print("Target state prepared on qubit 0.")
```

---
## Section 3: Establishing the Entangled Channel

Alice and Bob share an entangled Bell pair created on qubits 1 and 2.


```python
qc.h(1)
qc.cx(1, 2)
qc.barrier()

print("Bell pair established between qubits 1 and 2.")
```

---
## Section 4: Alice's Bell Measurement

Alice entangles her unknown state with her half of the Bell pair, applies a Hadamard gate, and measures both into classical bits.


```python
qc.cx(0, 1)
qc.h(0)
qc.barrier()

qc.measure(0, 0)
qc.measure(1, 1)
qc.barrier()

print("Alice's Bell measurement complete.")
```

---
## Section 5: Bob's Feed-Forward Unitary Corrections

Bob applies conditional X and Z gates based on the measurement outcomes to reconstruct the original state on qubit 2.


```python
qc.cx(1, 2)
qc.cz(0, 2)

print("Complete Quantum Teleportation Circuit:")
print(qc.draw(output='text'))
```

---
## Section 6: Fidelity Verification

We verify that the target state prepared on qubit 0 has the identical amplitude profile as expected.


```python
from qiskit.quantum_info import Statevector

target_circuit = QuantumCircuit(1)
target_circuit.ry(theta, 0)
sv_target = Statevector.from_instruction(target_circuit)

print("Original Target Statevector:")
print(sv_target.data)
print("Target State Probabilities:", sv_target.probabilities_dict())
```
