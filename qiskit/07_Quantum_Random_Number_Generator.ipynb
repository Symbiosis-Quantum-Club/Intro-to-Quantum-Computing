## Section 1: The 1-Bit Quantum Coin Toss

Measuring a qubit in equal superposition yields true non-deterministic randomness.


```python
from qiskit import QuantumCircuit
from qiskit.primitives import StatevectorSampler

def quantum_coin_toss():
    qc = QuantumCircuit(1)
    qc.h(0)
    qc.measure_all()
    
    sampler = StatevectorSampler()
    job = sampler.run([qc], shots=1)
    bit = list(job.result()[0].data.meas.get_counts().keys())[0]
    return "Heads (0)" if bit == '0' else "Tails (1)"

# Execute 5 sample tosses
for i in range(5):
    print(f"Toss {i + 1}: {quantum_coin_toss()}")
```

---
## Section 2: Multi-Bit Quantum Integer Sampling

By preparing $n$ qubits in superposition, we sample numbers uniformly in the range $[0, 2^n - 1]$.


```python
def sample_quantum_integer(num_bits=4):
    qc = QuantumCircuit(num_bits)
    for q in range(num_bits):
        qc.h(q)
    qc.measure_all()
    
    sampler = StatevectorSampler()
    job = sampler.run([qc], shots=1)
    bitstring = list(job.result()[0].data.meas.get_counts().keys())[0]
    return int(bitstring, 2)

sample_integers = [sample_quantum_integer(4) for _ in range(8)]
print("8 Sample 4-bit random integers (range 0 to 15):")
print(sample_integers)
```

---
## Section 3: Bounded 6-Sided Dice Roller

To generate a die roll (1 to 6), we generate 3-bit numbers (0 to 7) and reject values outside the target range.


```python
def roll_quantum_die():
    while True:
        val = sample_quantum_integer(3)
        if 1 <= val <= 6:
            return val

sample_rolls = [roll_quantum_die() for _ in range(12)]
print("Sample 12 die rolls:")
print(sample_rolls)
```

---
## Section 4: Statistical Uniformity Evaluation

We verify the uniform distribution of the generator across 500 samples.


```python
import matplotlib.pyplot as plt
import numpy as np

num_samples = 500
dice_results = [roll_quantum_die() for _ in range(num_samples)]

plt.figure(figsize=(6, 3))
plt.hist(dice_results, bins=np.arange(0.5, 7.5, 1), rwidth=0.8, color='#2ca02c', edgecolor='black')
plt.xlabel("Die Face Value")
plt.ylabel("Frequency")
plt.title(f"Quantum Die Roll Distribution ({num_samples} Trials)")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
