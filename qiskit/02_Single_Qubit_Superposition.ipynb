## Section 1: Complex Numbers and Probability Amplitudes

Quantum states are represented as complex amplitude vectors. For a single qubit:
$$|\psi\rangle = \alpha |0\rangle + \beta |1\rangle$$
The normalization constraint requires that $|\alpha|^2 + |\beta|^2 = 1$.


```python
import numpy as np

# Demonstrate normalization constraint with complex numbers
alpha = 1.0 / np.sqrt(2)
beta = 1.0 / np.sqrt(2)

prob_0 = np.abs(alpha)**2
prob_1 = np.abs(beta)**2
total_prob = prob_0 + prob_1

print(f"Alpha: {alpha:.4f} | Beta: {beta:.4f}")
print(f"P(0):  {prob_0:.4f} | P(1): {prob_1:.4f}")
print(f"Total Probability: {total_prob:.4f}")
```

---
## Section 2: Ground State Verification

Every freshly allocated qubit in Qiskit starts in the state $|0\rangle$. We verify this using the `Statevector` class from `qiskit.quantum_info`.


```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

# Initialize 1-qubit circuit with no gates
ground_qc = QuantumCircuit(1)

# Compute statevector
sv_ground = Statevector.from_instruction(ground_qc)

print("Ground Statevector Data:")
print(sv_ground.data)
print("Ground State Probabilities:", sv_ground.probabilities_dict())
```

---
## Section 3: Applying the Hadamard Gate

The Hadamard gate creates an equal superposition of $|0\rangle$ and $|1\rangle$:
$$H|0\rangle = \frac{|0\rangle + |1\rangle}{\sqrt{2}} = |+\rangle$$


```python
# Apply Hadamard to qubit 0
plus_qc = QuantumCircuit(1)
plus_qc.h(0)

print(plus_qc.draw(output='text'))
```

---
## Section 4: Analytical Statevector Inspection

We compute the resulting statevector and extract exact probability values.


```python
# Compute statevector for |+> state
sv_plus = Statevector.from_instruction(plus_qc)

print("Superposition Statevector (|plus>):")
print(sv_plus.data)

# Extract probabilities dictionary
probs_plus = sv_plus.probabilities_dict()
print("\nExact Measurement Probabilities:")
for basis_state, probability in probs_plus.items():
    print(f"State |{basis_state}>: {probability:.4f} ({probability * 100:.1f}%)")
```

---
## Section 5: The Orthogonal Superposition State: $|-\rangle$

Applying an X gate before the Hadamard gate produces the orthogonal superposition state:
$$H|1\rangle = \frac{|0\rangle - |1\rangle}{\sqrt{2}} = |-\rangle$$
Notice that while the measurement probabilities are identical ($50\% / 50\%$), the relative phase is negative!


```python
minus_qc = QuantumCircuit(1)
minus_qc.x(0)
minus_qc.h(0)

sv_minus = Statevector.from_instruction(minus_qc)
print("Statevector (|minus>):")
print(sv_minus.data)
print("Probabilities:", sv_minus.probabilities_dict())
```

---
## Section 6: Reversibility of Quantum Transformations

All quantum gate operations (except measurement) are unitary and strictly reversible. Applying the Hadamard gate twice returns the qubit to its initial state:
$$H \cdot H = I$$


```python
reversal_qc = QuantumCircuit(1)
reversal_qc.h(0)
reversal_qc.h(0)

sv_reversed = Statevector.from_instruction(reversal_qc)
print("Circuit with H applied twice:")
print(reversal_qc.draw(output='text'))
print("\nFinal state after H * H:", sv_reversed.data)
print("Probabilities after reversal:", sv_reversed.probabilities_dict())
```
