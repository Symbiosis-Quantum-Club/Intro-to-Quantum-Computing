## Section 1: Graph Definition and QAOA Circuit Synthesis


```python
import networkx as nx
import numpy as np
from qiskit import QuantumCircuit

G = nx.Graph()
G.add_edges_from([(0, 1), (1, 2), (2, 3), (3, 0), (0, 2)])

def build_qaoa_circuit(gamma=0.5, beta=0.3):
    qc = QuantumCircuit(4)
    qc.h(range(4))
    
    # Cost layer
    for u, v in G.edges():
        qc.rzz(2 * gamma, u, v)
        
    # Mixer layer
    for q in range(4):
        qc.rx(2 * beta, q)
        
    qc.measure_all()
    return qc

sample_circuit = build_qaoa_circuit()
print("QAOA Circuit (Single Layer):")
print(sample_circuit.draw(output='text'))
```
