# Notebook 10: Simulating Noisy Quantum Channels

Real quantum processors experience environmental decoherence. This notebook simulates noisy quantum channels using density matrices.

---

## Learning Objectives
1. Model depolarizing and damping channels.
2. Simulate mixed states using `cirq.DensityMatrixSimulator`.


---
## Real-World Applications & Modern Use Cases

Noise modeling and density matrix simulations are crucial for:
- **Realistic NISQ Algorithm Benchmarking:** Predicting how algorithmic performance degrades under physical thermal relaxation ($T_1$) and dephasing ($T_2$).
- **Quantum Error Mitigation (QEM):** Calibrating zero-noise extrapolation (ZNE) and probabilistic error cancellation (PEC) models to recover noise-free observable estimates.


```python
import cirq

q = cirq.LineQubit(0)

circuit = cirq.Circuit(
    cirq.X(q),
    cirq.depolarize(p=0.15)(q)
)

print("Noisy Circuit:")
print(circuit)

sim = cirq.DensityMatrixSimulator()
res = sim.simulate(circuit)
print("\nSimulated Density Matrix:")
print(res.final_density_matrix)
```
