## Section 1: Importing Cirq and Verifying Version


```python
import cirq
print("Cirq Framework Version:", cirq.__version__)
```

---
## Section 2: Qubit Types in Cirq

Cirq classifies qubits according to their topological constraints:
1. `NamedQubit`: Abstract labeled identifiers for algorithm design.
2. `LineQubit`: 1-dimensional array topologies.
3. `GridQubit`: 2-dimensional lattice coordinates matching Google's Sycamore architecture.


```python
# Define abstract named qubits
q_alice = cirq.NamedQubit("Alice")
q_bob = cirq.NamedQubit("Bob")

# Define 1D linear qubits
line_qubits = cirq.LineQubit.range(3)

# Define 2D grid qubits (row, column)
grid_qubit_1 = cirq.GridQubit(4, 5)
grid_qubit_2 = cirq.GridQubit(4, 6)

print("Named Qubits:", q_alice, q_bob)
print("Line Qubits:", line_qubits)
print("Grid Qubits:", grid_qubit_1, grid_qubit_2)
```

---
## Section 3: Creating Quantum Operations

In Cirq, applying a gate to a qubit produces an `Operation`. Operations are the atomic instructions inside a circuit.


```python
q = cirq.LineQubit(0)

# Create individual operations
op_h = cirq.H(q)
op_x = cirq.X(q)
op_measure = cirq.measure(q, key='result')

print("Operation 1:", op_h)
print("Operation 2:", op_x)
print("Operation 3:", op_measure)
```

---
## Section 4: Assembling and Printing the Circuit


```python
circuit = cirq.Circuit()
circuit.append(op_h)
circuit.append(op_x)
circuit.append(op_measure)

print("Complete Cirq Circuit Diagram:")
print(circuit)
```

---
## Section 5: Inspecting Circuit Structure

We inspect the operation count and the set of qubits involved in the circuit.


```python
print(f"Total Operations: {len(list(circuit.all_operations()))}")
print(f"Active Qubits:    {circuit.all_qubits()}")
```
