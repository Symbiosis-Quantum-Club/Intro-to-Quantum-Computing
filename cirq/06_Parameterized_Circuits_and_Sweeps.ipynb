# Notebook 06: Parameterized Circuits and SymPy Sweeps

Variational quantum algorithms require circuits with symbolic parameters evaluated across multiple values.

---

## Learning Objectives
1. Define symbolic circuit parameters using `sympy.Symbol`.
2. Execute parameter sweeps with `cirq.Linspace`.
3. Evaluate probability curves over angle ranges.


---
## Real-World Applications & Modern Use Cases

Symbolic parameter sweeps are standard practice in:
- **Variational Algorithm Hyperparameter Sweeps:** Mapping energy landscapes in VQE and QAOA across multi-dimensional angle grids to avoid local minima.
- **Batched Cloud QPU Execution:** Compiling circuit DAGs once with symbolic parameters and uploading array tables to Google Quantum Engine, drastically eliminating compilation latency over network APIs.


```python
import cirq
import sympy

q = cirq.LineQubit(0)
theta = sympy.Symbol('theta')

circuit = cirq.Circuit(
    cirq.rx(theta)(q),
    cirq.measure(q, key='m')
)

print("Parameterized Circuit:")
print(circuit)
```

---
## Section 1: Executing the Parameter Sweep


```python
sweep = cirq.Linspace(key='theta', start=0, stop=3.14159, length=5)
sim = cirq.Simulator()
results = sim.run_sweep(circuit, params=sweep, repetitions=100)

print("Parameter Sweep Execution Results:")
for res in results:
    val = res.params['theta']
    p1 = res.histogram(key='m')[1] / 100
    print(f"Theta: {val:.3f} rad | P(1): {p1:.2f}")
```
