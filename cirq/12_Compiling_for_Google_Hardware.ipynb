## Real-World Applications & Modern Use Cases

Target hardware compilation and gate synthesis are critical in:
- **QPU Native Gate Execution:** Converting abstract mathematical circuits into native gate primitives (such as Sycamore gates, $\sqrt{\text{iSWAP}}$, or CZ) supported by hardware control electronics.
- **Circuit Depth Reduction:** Running peephole optimization passes to cancel redundant single-qubit rotations and merge consecutive operations, preserving coherence.


```python
import cirq

q0, q1 = cirq.LineQubit.range(2)
circuit = cirq.Circuit(
    cirq.H(q0),
    cirq.CNOT(q0, q1)
)

print("Original Circuit:")
print(circuit)

compiled = cirq.optimize_for_target_gateset(
    circuit,
    gateset=cirq.CZTargetGateset()
)

print("\nCompiled Circuit (CZ Native):")
print(compiled)
```
