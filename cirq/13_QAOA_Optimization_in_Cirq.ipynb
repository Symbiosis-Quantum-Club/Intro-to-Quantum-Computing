## Real-World Applications & Modern Use Cases

The Quantum Approximate Optimization Algorithm (QAOA) is deployed in industrial research for:
- **Logistics and Supply Chain Routing:** Formulating vehicle routing, fleet scheduling, and warehouse inventory placement as quadratic unconstrained binary optimization (QUBO) problems.
- **Telecommunication Network Allocation:** Optimizing frequency assignments and packet routing in dense wireless communications networks.


```python
import cirq
import sympy

q0, q1, q2 = cirq.LineQubit.range(3)
gamma = sympy.Symbol('gamma')
beta = sympy.Symbol('beta')

circuit = cirq.Circuit(
    cirq.H.on_each(q0, q1, q2),
    cirq.ZZPowGate(exponent=gamma)(q0, q1),
    cirq.ZZPowGate(exponent=gamma)(q1, q2),
    cirq.XPowGate(exponent=beta).on_each(q0, q1, q2),
    cirq.measure(q0, q1, q2, key='m')
)

print("QAOA Circuit in Cirq:")
print(circuit)
```
