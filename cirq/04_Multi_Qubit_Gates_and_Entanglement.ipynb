## Real-World Applications & Modern Use Cases

Multi-qubit entanglement and Bell state generation serve as the baseline engine for:
- **Quantum Telecommunication & Key Distribution:** Distributing entangled photon or qubit pairs across network nodes for secure communications and quantum repeaters.
- **Quantum Sensing & Metrology:** Utilizing entangled GHZ states to surpass the standard quantum limit and achieve Heisenberg-limited measurement precision for electric and magnetic fields.


```python
import cirq

q0, q1 = cirq.LineQubit.range(2)
bell_circuit = cirq.Circuit(
    cirq.H(q0),
    cirq.CNOT(q0, q1)
)

print("Bell State Circuit:")
print(bell_circuit)
```

---
## Section 1: Pure-State Simulation


```python
sim = cirq.Simulator()
result = sim.simulate(bell_circuit)

print("Simulated Wavefunction (Dirac Notation):")
print(result.dirac_notation())

print("\nStatevector amplitudes:")
print(result.final_state_vector)
```
