This notebook implements Grover's search algorithm targeting the computational basis state $|11\rangle$.

## Real-World Applications & Modern Use Cases

Amplitude amplification and Grover search primitives are applied to:
- **Cryptanalytic Strength Assessment:** Quantifying the quadratic reduction in effective key length for symmetric ciphers (e.g. evaluating AES-256 resilience against 128-bit quantum search).
- **Constraint Satisfaction & Combinatorics:** Accelerating combinatorial search routines for graph coloring, boolean satisfiability (SAT), and route planning.


```python
import cirq

q0, q1 = cirq.LineQubit.range(2)

circuit = cirq.Circuit(
    cirq.H.on_each(q0, q1),
    cirq.CZ(q0, q1),
    cirq.H.on_each(q0, q1),
    cirq.X.on_each(q0, q1),
    cirq.CZ(q0, q1),
    cirq.X.on_each(q0, q1),
    cirq.H.on_each(q0, q1),
    cirq.measure(q0, q1, key='result')
)

print("Grover Circuit:")
print(circuit)

sim = cirq.Simulator()
res = sim.run(circuit, repetitions=500)
print("\nMeasurement counts (Target state 11):")
print(res.histogram(key='result'))
```
