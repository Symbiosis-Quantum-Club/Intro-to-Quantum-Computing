# Notebook 09: Quantum Fourier Transform in Cirq

This notebook implements the Quantum Fourier Transform (QFT) and demonstrates automated circuit inversion.

---

## Learning Objectives
1. Synthesize an $n$-qubit QFT in Cirq.
2. Invert operations using `cirq.inverse`.


---
## Real-World Applications & Modern Use Cases

The Quantum Fourier Transform (QFT) is an indispensable core subroutine in:
- **Quantum Phase Estimation (QPE):** Calculating molecular ground states and reaction barriers in advanced catalytic chemistry.
- **Shor's Factoring Algorithm:** Providing the exponential speedup in period-finding necessary to analyze RSA and discrete logarithm cryptographic schemes.


```python
import cirq

def build_cirq_qft(qubits):
    c = cirq.Circuit()
    n = len(qubits)
    for i in range(n):
        c.append(cirq.H(qubits[i]))
        for j in range(i + 1, n):
            power = 1.0 / (2 ** (j - i))
            c.append(cirq.CZ(qubits[j], qubits[i])**power)
    return c

qs = cirq.LineQubit.range(3)
qft = build_cirq_qft(qs)
print("QFT Circuit:")
print(qft)

inv_qft = cirq.inverse(qft)
print("\nInverse QFT Circuit:")
print(inv_qft)
```
