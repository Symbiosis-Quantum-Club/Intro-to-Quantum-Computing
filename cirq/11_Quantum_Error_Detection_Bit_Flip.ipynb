# Notebook 11: Quantum Error Detection and the 3-Qubit Code

This notebook demonstrates the 3-qubit repetition code for detecting and correcting single bit-flip errors.

---

## Learning Objectives
1. Encode logical qubits into physical registers.
2. Extract error syndromes using ancilla qubits.


---
## Real-World Applications & Modern Use Cases

Syndrome measurement and error detection codes are the foundation of:
- **Fault-Tolerant Quantum Architecture:** Paving the way toward topological surface codes and color codes that protect quantum information during long computations.
- **Autonomous Error Detection in Hardware:** Demonstrating logical error rates lower than underlying physical component error rates (break-even milestone).


```python
import cirq

data = cirq.LineQubit.range(3)
syndrome = [cirq.NamedQubit("s0"), cirq.NamedQubit("s1")]

circuit = cirq.Circuit(
    cirq.H(data[0]),
    cirq.CNOT(data[0], data[1]),
    cirq.CNOT(data[0], data[2]),
    # Inject error on data[1]
    cirq.X(data[1]),
    # Syndrome measurement
    cirq.CNOT(data[0], syndrome[0]),
    cirq.CNOT(data[1], syndrome[0]),
    cirq.CNOT(data[1], syndrome[1]),
    cirq.CNOT(data[2], syndrome[1]),
    cirq.measure(syndrome[0], syndrome[1], key='syndrome')
)

print("Bit-Flip Detection Circuit:")
print(circuit)
```
