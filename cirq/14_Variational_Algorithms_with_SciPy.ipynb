This notebook connects Cirq circuits with classical optimization routines.

## Real-World Applications & Modern Use Cases

Hybrid quantum-classical loops with classical optimizers (like COBYLA/Nelder-Mead) are used for:
- **Molecular Ground State Energy Estimation:** Determining equilibrium bond lengths and potential energy surfaces for battery electrolytes and solar cells.
- **NISQ Resilience:** Offloading numerical optimization to classical processors, leaving only parameter evaluation to quantum hardware where coherent state lifetime is limited.


```python
import cirq
import numpy as np
from scipy.optimize import minimize

q = cirq.LineQubit(0)

def energy_function(params):
    theta = params[0]
    c = cirq.Circuit(cirq.ry(theta)(q))
    sim = cirq.Simulator()
    state = sim.simulate(c).final_state_vector
    # Target Hamiltonian Z expectation value = |c0|^2 - |c1|^2
    return float(np.abs(state[0])**2 - np.abs(state[1])**2)

res = minimize(energy_function, [0.2], method='COBYLA')
print(f"Optimal angle found: {res.x[0]:.4f} rad")
print(f"Minimum energy found: {res.fun:.4f}")
```
