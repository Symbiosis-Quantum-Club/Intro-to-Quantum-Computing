## Real-World Applications & Modern Use Cases

Random circuit sampling and Cross-Entropy Benchmarking (XEB) are at the forefront of:
- **Quantum Computational Advantage:** Experimentally proving that a QPU can sample from probability distributions that are classically intractable for world-leading supercomputers.
- **Holistic QPU Characterization:** Providing system-wide fidelity metrics that capture simultaneous two-qubit errors, cross-talk, and control drift across the entire processor.


```python
import cirq

qubits = cirq.LineQubit.range(4)
circuit = cirq.Circuit()

# Layer of single-qubit operations
circuit.append([cirq.H(q) for q in qubits])
# Layer of entangling two-qubit operations
circuit.append([cirq.CZ(qubits[0], qubits[1]), cirq.CZ(qubits[2], qubits[3])])
circuit.append(cirq.measure(*qubits, key='m'))

print("Benchmark Circuit:")
print(circuit)

sim = cirq.Simulator()
res = sim.run(circuit, repetitions=200)
print("\nSampled bitstring distribution (first 5):")
print(res.data.head())
```
