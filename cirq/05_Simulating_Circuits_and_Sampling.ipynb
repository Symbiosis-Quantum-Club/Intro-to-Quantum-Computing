# Notebook 05: Simulation versus Hardware Sampling

Cirq provides `simulate()` for exact wavefunction calculations and `run()` for shot-based statistical sampling.

---

## Learning Objectives
1. Compute pure-state wavefunctions with `sim.simulate()`.
2. Sample repeated measurement outcomes with `sim.run()`.
3. Process result histograms and pandas DataFrames.


---
## Real-World Applications & Modern Use Cases

Distinguishing between statevector simulation and shot-based sampling is essential for:
- **Hardware Pre-flight Validation:** Verifying algorithm correctness against exact wavefunctions before expending physical QPU compute credits.
- **Shot-Noise Analysis:** Determining the statistical sampling budget (number of repetitions) required to resolve small expectation value shifts in variational algorithms under finite measurement noise.


```python
import cirq

q0, q1 = cirq.LineQubit.range(2)
circuit = cirq.Circuit(
    cirq.H(q0),
    cirq.CNOT(q0, q1),
    cirq.measure(q0, key='m0'),
    cirq.measure(q1, key='m1')
)

print("Measurement Circuit:")
print(circuit)
```

---
## Section 1: Executing 1,000 Repetitions


```python
sim = cirq.Simulator()
result = sim.run(circuit, repetitions=1000)

print("Measurement Histogram (m0):")
print(result.histogram(key='m0'))
print("Measurement Histogram (m1):")
print(result.histogram(key='m1'))

print("\nTabular DataFrame Sample (first 5 shots):")
print(result.data.head())
```
