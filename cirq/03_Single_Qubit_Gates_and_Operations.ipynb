## Real-World Applications & Modern Use Cases

Continuous single-qubit rotations and fractional powers are critical for:
- **Microwave Pulse Control:** Exponentiated Pauli gates directly correspond to shaped microwave drive pulses used to control superconducting transmon transitions.
- **Randomized Benchmarking (RB):** Decomposing random Clifford group elements into fractional elementary gates to quantify hardware gate fidelities above 99.9%.


```python
import cirq
import numpy as np

q = cirq.LineQubit(0)
print("Qubit allocated.")
```

---
## Section 1: Fractional Powers of Quantum Gates

In Cirq, any gate can be exponentiated using standard Python operator overloading (`**`):
- `cirq.X**1` is the standard Pauli-X (NOT) gate.
- `cirq.X**0.5` is the square root of NOT ($\sqrt{X}$).
- `cirq.Z**0.25` is the fourth root of Pauli-Z ($T$ gate).


```python
circuit = cirq.Circuit(
    cirq.X(q)**0.5,
    cirq.Y(q)**0.5,
    cirq.Z(q)**0.25
)

print("Circuit with Fractional Powers:")
print(circuit)
```

---
## Section 2: Calculating Unitary Matrices


```python
u_sqrt_x = cirq.unitary(cirq.X**0.5)
print("Unitary Matrix for X**0.5 (sqrt-NOT):")
print(np.round(u_sqrt_x, 3))

u_circuit = cirq.unitary(circuit)
print("\nTotal Circuit Unitary Matrix:")
print(np.round(u_circuit, 3))
```
